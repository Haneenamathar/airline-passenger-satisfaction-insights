## Hypothesis Testing

### Hypothesis Test 1 — Type of Travel and Passenger Satisfaction

During exploratory analysis, satisfaction levels differed noticeably between business and personal travel passengers. A Chi-square test of independence is used to determine whether this observed relationship is statistically significant.

**Null Hypothesis (H₀):** Type of travel and passenger satisfaction are independent; there is no statistically significant association between them.

**Alternative Hypothesis (H₁):** Type of travel and passenger satisfaction are associated; there is a statistically significant relationship between them.

**Significance level:** α = 0.05

A Chi-square test of independence is appropriate because both `Type of Travel` and `satisfaction` are categorical variables.

#### Observed Frequencies

Create a contingency table showing the number of satisfied and neutral or dissatisfied passengers within each type of travel.

```python
travel_contingency = pd.crosstab(
    df["Type of Travel"],
    df["satisfaction"]
)

travel_contingency
```

#### Chi-square Test of Independence

Calculate the Chi-square statistic and p-value to determine whether the association between type of travel and passenger satisfaction is statistically significant.

```python
chi2, p_value, dof, expected = stats.chi2_contingency(
    travel_contingency
)

print(f"Chi-square statistic: {chi2:.2f}")
print(f"P-value: {p_value:.6f}")
print(f"Degrees of freedom: {dof}")
print(f"Minimum expected frequency: {expected.min():.2f}")
```

#### Result and Interpretation

The Chi-square test produced a statistic of 20,882.22 with 1 degree of freedom. The p-value is less than 0.001, which is below the significance level of 0.05.

The minimum expected frequency is 13,924.51 which is well above 5, indicating that the expected-frequency assumption for the Chi-square test is satisfied.

Therefore, the **null hypothesis (H₀) is rejected**.

There is statistically significant evidence of an association between **type of travel and passenger satisfaction**. The observed frequencies show that business travellers are considerably more likely to be satisfied than personal travellers.

This supports the pattern identified during exploratory data analysis and suggests that type of travel is an important variable to consider when analysing and later predicting passenger satisfaction.

---

#### Effect Size — Cramér's V

Calculate Cramér's V to measure the strength of the association between type of travel and passenger satisfaction.

```python
import numpy as np

n = travel_contingency.to_numpy().sum()

cramers_v = np.sqrt(
    chi2 / (
        n * min(
            travel_contingency.shape[0] - 1,
            travel_contingency.shape[1] - 1
        )
    )
)

print(f"Cramér's V: {cramers_v:.3f}")
```

#### Effect Size Interpretation

Cramér's V is 0.449, indicating a moderate to relatively strong association between "type of travel and passenger satisfaction".

This supports the Chi-square test result by showing that the relationship is not only statistically significant but also meaningful in practical terms.

Therefore, type of travel appears to be an important characteristic associated with passenger satisfaction and may provide useful predictive information in later modelling.

## Core Statistical Concepts

### Mean

The mean is the average value of a numerical variable. It is calculated by adding all values and dividing by the number of observations. In this project, the mean is useful for understanding typical values such as passenger age, flight distance and flight delays. However, the mean can be influenced by unusually high or low values.

### Median

The median is the middle value when observations are arranged in order. It is less affected by extreme values than the mean. This is particularly useful for the flight delay variables, where the median delay is 0 minutes while the mean is higher because some passengers experience very large delays.

### Standard Deviation

Standard deviation measures how spread out observations are around the mean. A larger standard deviation indicates greater variability, while a smaller standard deviation indicates that observations are more concentrated around the mean.

### Hypothesis Testing

Hypothesis testing is used to determine whether evidence from the data is strong enough to reject a null hypothesis. In this project, a significance level of **0.05** is used. A p-value below 0.05 provides evidence to reject the null hypothesis.

### Basic Probability

Probability describes the likelihood of an event occurring and ranges from 0 to 1. In this dataset, the probability of selecting a satisfied passenger can be estimated by dividing the number of satisfied passengers by the total number of passengers.

---

### Probability of Passenger Satisfaction

Calculate the probability that a randomly selected passenger from the dataset is satisfied.

```python
satisfied_count = (df["satisfaction"] == "satisfied").sum()
total_passengers = len(df)

satisfaction_probability = satisfied_count / total_passengers

print(f"Satisfied passengers: {satisfied_count}")
print(f"Total passengers: {total_passengers}")
print(f"Probability of satisfaction: {satisfaction_probability:.4f}")
print(f"Percentage satisfied: {satisfaction_probability * 100:.2f}%")
```

### Probability Interpretation

The probability of randomly selecting a satisfied passenger from the dataset is approximately 0.433, or 43.3%.

This means that fewer than half of the passengers in the cleaned dataset are satisfied, which is consistent with the satisfaction distribution identified during exploratory data analysis.
